# SQL EDA aggregations

## Table View 1: v_cohort

Aggregation to view 
1) ICU Length of Stay 
2) Hospital Length of Stay
3) Age at admission 

```python
# ========================================
# Table View 1: v_cohort
# =========================================
# Can join icustays with admission through 'hadm_id'
# Can join icustays with patients through 'subject_id'

# From icustays need columns: [intime, outtime] 
#   To determine length of stay in ICU
# FROM admissions need columns: [admittime, dischtime]
#   To determine hospital length of stay
# FROM patients need columns: [dob]
#   determine the age of patients using admittime
pd.read_sql(
    """
    SELECT 
        p.subject_id, p.dob,
        i.intime, i.outtime, 
        a.admittime, a.dischtime,
        
        ROUND((JULIANDAY(outtime) - JULIANDAY(intime)) * 24, 2) AS icu_los_hours,
        ROUND(JULIANDAY(dischtime) - JULIANDAY(admittime), 2) AS hosp_los_days,
        ROUND((JULIANDAY(admittime) - JULIANDAY(dob)) / 365, 1) AS age_at_admit
    FROM icustays AS i 
    JOIN admissions AS a ON i.hadm_id = a.hadm_id
    JOIN patients AS p ON a.subject_id = p.subject_id
    """, 
    conn
)
# Note: There are some people about 300 years old. This is okay since this data was modified by PhysioNet to be untraceable/deidentified.
# The SQL query correctly calculates the difference for DOB
```

## Table View 2: v_vitals_24h

In order to create Table View 2 we need to know what each itemid is. This is stored in d_items. 

```python
pd.read_sql("""
    SELECT itemid, label, category, unitname, dbsource
    FROM d_items
    WHERE label LIKE '%heart rate%'
        OR label LIKE '%blood pressure%'
        OR label LIKE '%SpO2%'
        OR label LIKE '%oxygen%'
        OR label LIKE '%temperature%'
        OR label LIKE '%GCS%'
        OR label LIKE '%respiratory%'
    ORDER BY dbsource, label
""", conn)
```

Further refine the category column to view 'routine vital signs'

```python
pd.read_sql("""
    SELECT itemid, label, category, unitname, dbsource
    FROM d_items
    WHERE category LIKE '%Routine%'
    ORDER BY dbsource, label
""", conn)
```

```python
pd.read_sql("""
    SELECT itemid, label, category, unitname
    FROM d_items
    WHERE category LIKE '%Routine%'
    AND (
        label LIKE '%heart rate%'
        OR label LIKE '%blood pressure%'
        OR label LIKE '%SpO2%'
        OR label LIKE '%temperature%'
        OR label LIKE '%respiratory%'
        OR label LIKE '%GCS%'
    )
    ORDER BY dbsource, label
""", conn)
```
