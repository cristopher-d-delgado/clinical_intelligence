```python
import sqlite3
import pandas as pd

pd.set_option('display.max_colwidth', None)  # Show full content of columns
pd.set_option('display.max_rows', None)      # Show all rows
```

# Determine DB schema

```python
# Connect to sql db 
conn = sqlite3.connect("../db/mimic.db")

# Determine what tables Exist 
pd.read_sql(
    """
    SELECT name 
    FROM sqlite_master
    WHERE type='table' 
    ORDER BY name
    """,
    conn
)
```

```python
# As it can be tedious to write all the names this is a query generator. Now we just copy paste into the actual query we want
print(pd.read_sql(
    """
    SELECT 'SELECT ''' || name || ''' AS table_name, COUNT(*) AS row_count FROM ' || name || ' UNION ALL'
    FROM sqlite_master 
    WHERE type='table'; 
    """, 
    conn
))
```

```python
# View row counts of each Table (Should only be 100 Patients as its the DEMO Version of MIMIC 3)
pd.read_sql(
    """
    SELECT 'patients' AS table_name, COUNT(*) AS row_count FROM patients UNION ALL
    SELECT 'admissions' AS table_name, COUNT(*) AS row_count FROM admissions UNION ALL
    SELECT 'icustays' AS table_name, COUNT(*) AS row_count FROM icustays UNION ALL
    SELECT 'chartevents' AS table_name, COUNT(*) AS row_count FROM chartevents UNION ALL
    SELECT 'labevents' AS table_name, COUNT(*) AS row_count FROM labevents UNION ALL
    SELECT 'outputevents' AS table_name, COUNT(*) AS row_count FROM outputevents UNION ALL
    SELECT 'inputevents_cv' AS table_name, COUNT(*) AS row_count FROM inputevents_cv UNION ALL
    SELECT 'inputevents_mv' AS table_name, COUNT(*) AS row_count FROM inputevents_mv UNION ALL
    SELECT 'prescriptions' AS table_name, COUNT(*) AS row_count FROM prescriptions UNION ALL
    SELECT 'procedures_icd' AS table_name, COUNT(*) AS row_count FROM procedures_icd UNION ALL
    SELECT 'diagnoses_icd' AS table_name, COUNT(*) AS row_count FROM diagnoses_icd UNION ALL
    SELECT 'd_icd_diagnoses' AS table_name, COUNT(*) AS row_count FROM d_icd_diagnoses UNION ALL
    SELECT 'd_icd_procedures' AS table_name, COUNT(*) AS row_count FROM d_icd_procedures UNION ALL
    SELECT 'd_items' AS table_name, COUNT(*) AS row_count FROM d_items UNION ALL
    SELECT 'd_labitems' AS table_name, COUNT(*) AS row_count FROM d_labitems UNION ALL
    SELECT 'microbiologyevents' AS table_name, COUNT(*) AS row_count FROM microbiologyevents UNION ALL
    SELECT 'datetimeevents' AS table_name, COUNT(*) AS row_count FROM datetimeevents UNION ALL
    SELECT 'callout' AS table_name, COUNT(*) AS row_count FROM callout UNION ALL
    SELECT 'caregivers' AS table_name, COUNT(*) AS row_count FROM caregivers UNION ALL
    SELECT 'services' AS table_name, COUNT(*) AS row_count FROM services UNION ALL
    SELECT 'transfers' AS table_name, COUNT(*) AS row_count FROM transfers UNION ALL
    SELECT 'cptevents' AS table_name, COUNT(*) AS row_count FROM cptevents UNION ALL
    SELECT 'drgcodes' AS table_name, COUNT(*) AS row_count FROM drgcodes
    """,
    conn
)
```

```python
# View each table for 5 rows 
tables = conn.execute(
    "SELECT name FROM sqlite_master WHERE type='table' ORDER BY name"
).fetchall()

for (t,) in tables:
    print(f"\n{'='*60}")
    print(f" {t.upper()}")
    print(f"{'='*60}")
    display(pd.read_sql(f"SELECT * FROM {t} LIMIT 5", conn))
```

```python
# Column info for every table
tables = conn.execute(
    "SELECT name FROM sqlite_master WHERE type='table' ORDER BY name"
).fetchall()

for (t,) in tables:
    print(f"\n{'='*60}")
    print(f"  {t.upper()}")
    print(f"{'='*60}")
    display(pd.read_sql(f"PRAGMA table_info({t})", conn))
```
